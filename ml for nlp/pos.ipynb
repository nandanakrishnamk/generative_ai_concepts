```python
import nltk
```

```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

stemmer=PorterStemmer()

para='''Students are studying various subjects at universities while researchers are analyzing data and developing innovative technologies. Several organizations are working on improving healthcare systems, reducing environmental pollution, and creating sustainable solutions for future generations. People communicate through different platforms, sharing ideas, learning new skills, and collaborating on projects. Advances in science and engineering are transforming industries, enabling businesses to operate more efficiently and helping communities solve complex problems.'''
stop=stopwords.words('english')

sentences=nltk.sent_tokenize(para)

#stemming after removing stop words

for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    #sentences[i]=' '.join(words)
    pos_tag1=nltk.pos_tag(words)

    print(pos_tag1)
```
